```python
raiz = Path.cwd()
nome_arquivo = "cnes_estabelecimentos_MA_2024_completo.csv"

# 2. Busca o arquivo automaticamente na pasta 'dados', na raiz ou em qualquer subpasta do projeto
caminhos_possiveis = list(raiz.rglob(nome_arquivo))

if caminhos_possiveis:
    caminho_csv = caminhos_possiveis[0]
    print(f"Arquivo encontrado em: {caminho_csv}")
    df = pd.read_csv(caminho_csv, low_memory=False)
else:
    raise FileNotFoundError(f"O arquivo '{nome_arquivo}' não está dentro da pasta do projeto ({raiz}).")
```

```python
# 1. Dicionário de mapeamento
mapa_colunas = {
    # Identificação e Documentos
    'CNES': 'cnes',
    'CO_CNES': 'cnes',
    'NO_FANTASIA': 'nome_fantasia',
    'NO_RAZAO_SOCIAL': 'razao_social',
    'CPF_CNPJ': 'cpf_cnpj',
    'PF_PJ': 'tipo_pessoa',
    'CNPJ_MAN': 'cnpj_mantenedora',
    
    # Classificação e Gestão
    'TP_UNIDADE': 'tipo_unidade',
    'TP_UNID': 'tipo_unidade',
    'VINC_SUS': 'vinculo_sus',
    'TPGESTAO': 'tipo_gestao',
    'TP_GESTAO': 'tipo_gestao',
    'NIV_DEP': 'nivel_dependencia',
    'CLIENTEL': 'perfil_clientela',
    'TURNO_AT': 'turno_atendimento',
    'COMPETEN': 'competencia_ano_mes',
    
    # Localização e Endereço
    'CODUFMUN': 'codigo_municipio_ibge',
    'CO_MUNICIPIO': 'codigo_municipio_ibge',
    'munResNome': 'municipio_nome',
    'munResUf': 'municipio_uf',
    'munResTipo': 'municipio_tipo',
    'REGSAUDE': 'regiao_saude',
    'COD_CEP': 'cep',
    'CO_CEP': 'cep',
    'DS_LOGRADOURO': 'logradouro',
    'NO_LOGRADOURO': 'logradouro',
    'NU_ENDERECO': 'numero',
    'NO_BAIRRO': 'bairro',
    'NU_TELEFONE': 'telefone',
    
    # Georreferenciamento (Essencial para o PostGIS / Mapas)
    'munResLat': 'municipio_latitude',
    'munResLon': 'municipio_longitude',
    'munResAlt': 'municipio_altitude',
    'munResArea': 'municipio_area_km2',
    'LATITUDE': 'latitude',
    'NU_LATITUDE': 'latitude',
    'LONGITUDE': 'longitude',
    'NU_LONGITUDE': 'longitude'
}

# 2. Filtrar apenas as colunas do dicionário que realmente existem no seu DataFrame
colunas_para_manter = [col for col in mapa_colunas.keys() if col in df.columns]

# 3. Criar df1 mantendo somente essas colunas e aplicando os novos nomes
df1 = df[colunas_para_manter].rename(columns=mapa_colunas).copy()

print("Formato anterior:", df.shape)
print("Novo formato (apenas colunas mapeadas):", df1.shape)
print("Colunas finais:", list(df1.columns))
```

```python
# O campo 'COMPETEN' (formato YYYYMM, ex: 202401 até 202412) garante a ordem temporal
if 'COMPETEN' in df1.columns:
    df1 = df1.sort_values(by='COMPETEN')

#EXECUTAR A DEDUPLICAÇÃO (O Coração da Opção 1)
df2 = df1.drop_duplicates(subset=['cnes'], keep='last')


print(f"Linhas originais (12 meses): {len(df1):,}")
print(f"Linhas únicas (após deduplicação): {len(df2):,}")
```

```python
# Criar uma tabela com o total e a porcentagem de nulos por coluna
faltantes = pd.DataFrame({
    'Coluna': df2.columns,
    'Faltantes': df2.isnull().sum().values,
    'Porcentagem (%)': (df2.isnull().sum().values / len(df2) * 100).round(2)
})

# Ordenar pelas colunas que têm mais dados faltantes
faltantes_ordenado = faltantes.sort_values(by='Faltantes', ascending=False)

# Exibir todas as linhas
pd.set_option('display.max_rows', None)
faltantes_ordenado
```

```python
#Limpar as colunas que nao possuem nenhum dado
df3 = df2.dropna(how='all', axis=1)
```

```python
#Deixar no Dataframe só os Espaços que eu quero na coluna de TP_UNIDADE

#identificar a coluna que vai ser modificada
colunas_selecionadas =  'tipo_unidade'

#Identificar quais unidades eu quero manter
tipos_unidade = ['Centro de saúde / Unidade básica',
    'Posto de saúde',
    'Policlínica']
#Aplicar o filtro acima

df5 = df4[df4[colunas_selecionadas].isin(tipos_unidade)].copy()
```

```python
#Deduplicar (manter apenas o cadastro mais recente de cada CNES em 2024)
col_cnes = 'cnes' if 'cnes' in df5.columns else 'CO_CNES'
df6 = df5.drop_duplicates(subset=[col_cnes], keep='last')

print(f"Total de registros antes do filtro: {len(df):,}")
print(f"Total após filtrar os tipos escolhidos (12 meses): {len(df4):,}")
print(f"Total de estabelecimentos únicos filtrados: {len(df5):,}")
```

```python
# 1. Função para limpar e padronizar códigos sujos (ex: '001', '1', '2111', '008,' -> '01', '01', '11', '08')
def padronizar_regiao_saude(valor):
    if pd.isna(valor):
        return None
    numeros = re.findall(r'\d+', str(valor).strip())
    if not numeros:
        return None
    cod = int(numeros[0])
    # Corrige casos cadastrados com prefixo da UF 21 (ex: 2111 -> 11)
    if 2100 < cod < 2200:
        cod = cod % 100
    if 1 <= cod <= 18:
        return f"{cod:02d}"
    return None

# 2. Construir o mapa oficial (codigo_municipio_ibge -> regiao_saude) usando a moda da base completa 'df'
df['reg_limpa'] = df['REGSAUDE'].apply(padronizar_regiao_saude)
mapa_ibge_regiao = (
    df.dropna(subset=['reg_limpa'])
    .groupby('CODUFMUN')['reg_limpa']
    .agg(lambda x: x.mode().iloc[0])
    .to_dict()
)

# Adicionar manualmente o único município sem registro preenchido na base (210375 = Davinópolis -> Região 11 - Imperatriz)
mapa_ibge_regiao[210375] = '11'

# 3. Dicionário com o nome descritivo de cada Região de Saúde do Maranhão
nomes_regioes_ma = {
    '01': '01 - São Luís',
    '02': '02 - Rosário',
    '03': '03 - Itapecuru Mirim',
    '04': '04 - Chapadinha',
    '05': '05 - Codó',
    '06': '06 - Pinheiro',
    '07': '07 - Viana',
    '08': '08 - Santa Inês',
    '09': '09 - Zé Doca',
    '10': '10 - Açailândia',
    '11': '11 - Imperatriz',
    '12': '12 - Balsas',
    '13': '13 - São João dos Patos',
    '14': '14 - Presidente Dutra',
    '15': '15 - Pedreiras',
    '16': '16 - Barra do Corda',
    '17': '17 - Bacabal',
    '18': '18 - Caxias / Timon'
}

# 4. Aplicar o preenchimento padronizado no df6 e criar a coluna com o nome da região
df6 = df6.copy()
df6['regiao_saude'] = df6['codigo_municipio_ibge'].map(mapa_ibge_regiao)
df6['regiao_saude_nome'] = df6['regiao_saude'].map(nomes_regioes_ma)

```
